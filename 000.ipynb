```python
# functions
#
def add_nums(num1: int, num2: int) -> int:
    return num1 + num2

result: int = add_nums(10, 20)
print(f"Sum of 10 and 20 is: {result}")

def is_even(number: int) -> bool:
    return number % 2 == 0
check_num: int = 42
print(f"Is {check_num} even? {is_even(check_num)}")

user_num = int(input("Enter a number to check if it's even: "))
print(f"Is {user_num} even? {is_even(user_num)}")
```

```python
# objects and classes
#
class Bank_Account:
    def __init__(self, name: str, balance: float) -> None:
        self.name = name
        self.balance = balance

    def deposit(self, amount: float) -> None:
        self.balance += amount
        print(f"Deposited {amount}. New balance is {self.balance}.")

    def withdraw(self, amount: float) -> None:
        if amount > self.balance:
            print("Insufficient funds!")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance is {self.balance}.")

# use the Bank_Account class to create an object, that represents a user's bank account
account_10001 = Bank_Account("Metri", 4500.0)
account_10001.deposit(1500.0)
account_10001.withdraw(20000.0)
account_10001.withdraw(1500.0)
```

```python
# functions
# beginner - tic tac toe board
def print_board(board: list) -> None:
    for row in board:
        print(" | ".join(row))
        print("-" * 9)
tic_tac_toe_board = [["X", "O", "X"],
                      [" ", "X", " "],
                      ["O", " ", "O"]]
print_board(tic_tac_toe_board)

# empty space to separate the boards
print("\n" * 2)

# print blank board
def print_blank(board: list) -> None:
    for row in board:
        print(" | ".join(row))
        print("-" * 10)
blank_board = [[" ", " ", " "],
                [" ", " ", " "],
                [" ", " ", " "]]
print_blank(blank_board)
```

```python
# reverse a name function
def reverse_name(name: str) -> str:
    return name[::-1]
user_name: str = input("Enter your name to reverse it: ")
print(f"😂 {user_name} reversed is: {reverse_name(user_name)}")
```

```python
# python libraries and function
import random
def roll_dice(sides: int) -> int:
    return random.randint(1, sides)
dice_sides: int = 6
print(f"Rolling a {dice_sides}-sided dice: {roll_dice(dice_sides)}")
```

```python
# function to create a text file
def create_text_file(filename: str, location: str) -> None:
    with open(f"{location}/{filename}.txt", "w") as file:
        file.write("File created successfully!")
file_name: str = "new_file"
file_location: str = "."  # current directory
create_text_file(file_name, file_location)
print(f"Text file '{file_name}.txt' created at location: {file_location}")
```
